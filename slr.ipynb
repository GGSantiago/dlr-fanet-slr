# Systematic Literature Review

A Systematic Literature Review (SLR) is well defined and structured method or process for identifying, selecting and critically appraising literature on a specific topic.

7 steps are considered to conduct a SLR:

1. Define the resarch question
2. Define the scope, objectives and inclusion or exclusion criteria.
3. Search for studies in relevant databases.
4. Screen and select studies for detailed review using the scope, objectives, and criteria defined in Step 2.
5. Extract data from studies resulting from step 4.
6. Synthesize results.
7. Disseminate results.

This python Notebook steps 1 to 4 serves as reliable proof and as an example of the SLR conducted by Santiago García Gil during his PhD.

## Step 1: Define the research question

To define the research question the PICOC framework is used:
|     PICOC    	|  Definition                                                                                                          	|
|:------------:	|--------------------------------------------------------------------------------------------------------------------	|
| Population   	| Mesh/Ad Hoc networks where Unmanned Aerial Systems (UAS) are used as the platform of nodes.                        	|
| Intervention 	| Application of Deep Reinforcement Learning (DRL)                                                                   	|
| Comparison   	| No DRL or application of traditional methods.                                                                      	|
| Outcome      	| Impact in network-level metrics such as throughput, latency, jitter, packet loss, delivery ratio, and reliability. 	|
| Context      	| Computer networks                                                                                                  	|

## Step 2: Define the scope, objectives and inclusion or exclusion criteria
The thesis of Santiago García Gil is deeply related to 3 topics: ad hoc computer networks, UAS and DRL. Given these facts and the result of step 1, the scope of the SLR is defined as follows:

- Scope:
    - UAS platforms acting as nodes, relays, or routers in ad hoc network topologies.
    - Applications of Deep Reinforcement Learning for network control, routing, resource allocation, connectivity management, and QoS improvement.
    - Evaluation focused on network-level metrics such as throughput, latency, jitter, packet loss, delivery ratio, and reliability.
    - Context limited to computer networking research rather than UAV flight dynamics or perception.

- Objectives:
    1. Identify and categorize DRL methods applied to UAS-enabled ad hoc network problems.
    2. Study the usefulness given the reported performance improvements and QoS results across comparable approaches.

- Inclusion criteria (IC):
    1. (IC1) Peer-reviewed journal or conference papers written in English.
    2. (IC2) Studies published from 2015 to 2026 that apply DRL in aerial ad hoc networking scenarios.
    3. (IC3) Papers with experimental or simulation results on network performance or QoS.
    4. (IC4) Research with a clear network-level focus, not limited to UAV path planning, vision, or flight control.

- Exclusion criteria (EC):
    - (EC1) Works that use classical RL without deep learning components.
    - (EC2) UAV research focused only on navigation, mapping, or sensing without ad hoc network operations.
    - (EC3) Ad hoc networking studies without multihop aerial/UAV nodes.
    - (EC4) Survey papers, editorials, patents, or non-peer-reviewed reports without original results.
    - (EC5) Studies that do not report measurable network metrics or QoS outcomes.

- Screening:
    1. Data reconciliation from resulting articles from different academical databases (elimination of: duplicates, non journal or conference papers and/or retracted papers).
    2. Title/abstract screening. Artificial Intelligence (AI) assisted with human in the loop. Inclusion/exclusion criteria defined in Step 2 is used.
    3. Full-text screening of remaining papers to confirm DRL use, network focus, and reported QoS/network metrics.



## Step 3: Seach for studies in relevant databases
Well known computer and telecomunication databases are considered for the current SLR, namely: IEEE Xplore, Scopus (Elsevier) and ACM Digital Library.

The following query has been designed for use in the advance search functionality of [IEEE Xplore](https://ieeexplore.ieee.org/search/advanced/command):
```plaintext
("All Metadata":"mesh" OR "All Metadata":"ad hoc" OR "All Metadata":"vanet" OR "All Metadata":"fanet") 
    AND ("All Metadata":"network" OR "All Metadata":"networks" OR "All Metadata":"networking") 
    AND ("All Metadata":"drl" OR "All Metadata":"deep reinforcement learning" OR "All Metadata":"reinforcement learning")
    AND ("All Metadata":"uav" OR "All Metadata":"uas" OR "All Metadata":"drone" OR "All Metadata":"drones" ) 
    AND ("All Metadata":"latency" OR "All Metadata":"jitter" OR "All Metadata":"qos" OR "All Metadata":"quality of service" OR "All Metadata":"throughtput")
    AND NOT "Publication Title":"review" 
    AND NOT "Publication Title":"survey" 
```

Similarly, the following query is designed for [Scopus](https://www.scopus.com/search/form.uri?display=advanced):
```plaintext
(TITLE-ABS-KEY("mesh") OR TITLE-ABS-KEY("ad hoc") OR TITLE-ABS-KEY("vanet") OR TITLE-ABS-KEY("fanet")) 
    AND (TITLE-ABS-KEY("network") OR TITLE-ABS-KEY("networks") OR TITLE-ABS-KEY("networking")) 
    AND (TITLE-ABS-KEY("drl") OR TITLE-ABS-KEY("deep reinforcement learning") OR TITLE-ABS-KEY("reinforcement learning"))
    AND (TITLE-ABS-KEY("uav") OR TITLE-ABS-KEY("uas") OR TITLE-ABS-KEY("drone") OR TITLE-ABS-KEY("drones") ) 
    AND (TITLE-ABS-KEY("latency") OR TITLE-ABS-KEY("jitter") OR TITLE-ABS-KEY("qos") OR TITLE-ABS-KEY("quality of service") OR TITLE-ABS-KEY("throughtput"))
    AND NOT TITLE("review") 
    AND NOT TITLE("survey") 
```

As of 2026/05/15, the query yields 65 entries.

Finally, for [ACM Digital Library](https://dl.acm.org/search/advanced) the query is as follows:
```plaintext
AllField:(mesh OR "ad hoc" OR vanet OR fanet) 
    AND AllField:(network OR networks OR networking) 
    AND AllField:(drl OR "deep reinforcement learning" OR "reinforcement learning") 
    AND AllField:(uav OR uas OR drone OR drones) 
    AND AllField:(latency OR jitter OR qos OR "quality of service" OR "throughtput") 
    AND Title:(NOT survey) 
    AND Title:(NOT review) 
    AND Abstract:(NOT survey) 
    AND Abstract:(NOT review)
```
For ACM Digital Library, only results from 2018 onwards are considered.

As of 2026/05/15, queries yields 76, 65 and 938 entries, respectively.

### Step 4.1: Pruning and duplication removal

````python
import bibtexparser

with open('step_3_results/acm.bib') as bibtex_file:
   acm_database = bibtexparser.load(bibtex_file)
with open('step_3_results/ieee.bib') as bibtex_file:
   ieee_database = bibtexparser.load(bibtex_file)
with open('step_3_results/scopus.bib') as bibtex_file:
   scopus_database = bibtexparser.load(bibtex_file)   
````

````python
import pandas as pd

acm_df = pd.DataFrame(acm_database.entries)
ieee_df = pd.DataFrame(ieee_database.entries)
scopus_df = pd.DataFrame(scopus_database.entries)

print('Information about ACM Library query result')
print(f'No. of entries: {len(acm_df)}')
print(f'Columns: {len(acm_df.columns)}: {acm_df.columns.tolist()}')
print(f'Non-null values: {acm_df[['title', 'keywords', 'abstract']].count()}')
print()
print('Information about IEEE Xplore query result')
print(f'No. of entries: {len(ieee_df)}')
print(f'Columns: {len(ieee_df.columns)}: {ieee_df.columns.tolist()}')
print(f'Non-null values: {ieee_df[['title', 'keywords', 'abstract']].count()}')
print()
print('Information about Scopus query result')
print(f'No. of entries: {len(scopus_df)}')
print(f'Columns: {len(scopus_df.columns)}: {scopus_df.columns.tolist()}')
print(f'Non-null values: {scopus_df[['title', 'keywords', 'abstract']].count()}')
print()
````

````python
# Show a sample of entries with keywords and abstract

acm_df_filtered = acm_df[acm_df['abstract'].isnull() & acm_df['keywords'].isnull()]
print(f'No. of entries without keywords and abstract in ACM Library query result: {acm_df_filtered.shape[0]}')
print(acm_df_filtered['title'].sample(30))
print('-'*80)

# It seems that most of the entries in the ACM Library that do not have keywords nor abstract are proceedings of conferences.

acm_df_proceedings = acm_df[acm_df['title'].str.contains('Proceeding', case=False, na=False)]
print(f'No. of entries with "Proceeding" in the title in ACM Library query result: {acm_df_proceedings.shape[0]}')
print(acm_df_proceedings['title'].sample(30))
print('-'*80)

# Check the not intersecting entries
acm_df_filtered = pd.concat([acm_df_filtered, acm_df_proceedings], axis=0).drop_duplicates(keep=False)
print(f'No. of entries with keywords and abstract in ACM Library query result that are also proceedings of conferences: {acm_df_filtered.shape[0]}')
print(acm_df_filtered['title'].sample(30))
print('-'*80)

# It turns of that non intersecting entries are also proceedings of conferences that have abstracts.
# Remove all entries with "Proceeding" in the title from the ACM Library query result
acm_df_filtered = acm_df[~acm_df['title'].str.contains('Proceeding', case=False, na=False)]
print(f'No. of entries with keywords and abstract in ACM Library query result after removing proceedings of conferences: {acm_df_filtered.shape[0]}')
print(acm_df_filtered[['title', 'abstract', 'keywords']].count())

# Some nulls in the abstract column, but they are not proceedings of conferences. Let's check them.
acm_df_aux = acm_df_filtered[acm_df_filtered['abstract'].isnull() & acm_df_filtered['keywords'].isnull()]

print(acm_df_aux[['title', 'abstract', 'keywords']].sample(20))

# They seem to be entries of posters, demos and not useful for our SLR. Let's remove them.
acm_df_filtered = acm_df_filtered[~(acm_df_filtered['abstract'].isnull() & acm_df_filtered['keywords'].isnull())]
print(f'No. of entries with keywords and abstract in ACM Library query result after removing entries without keywords and abstract: {acm_df_filtered.shape[0]}')
print(acm_df_filtered[['title', 'abstract', 'keywords']].count())

acm_df_aux = acm_df_filtered[acm_df_filtered['abstract'].isnull()]
print(acm_df_aux['title'])
# They are unrelated old papers that are not relevant for our SLR. Let's remove them.
acm_df_filtered = acm_df_filtered[~acm_df_filtered.title.isin(acm_df_aux['title'])]
print(f'No. of clean entries in ACM Library query result: {acm_df_filtered.shape[0]}')
print(acm_df_filtered[['title', 'abstract', 'keywords']].count())
````

````python
acm_df_aux = acm_df_filtered[acm_df_filtered['abstract'].isnull() | acm_df_filtered['keywords'].isnull()][['title', 'abstract', 'keywords']]
print(acm_df_aux.count())
print(acm_df_aux[acm_df_aux['abstract'].isnull()])
````

````python
# Let's check if there are proceedings of conferences in the IEEE Xplore query result
ieee_df_proceedings = ieee_df[ieee_df['title'].str.contains('Proceedings', case=False, na=False)]
print(f'No. of entries with "Proceedings" in the title in IEEE Xplore query result: {ieee_df_proceedings.shape[0]}')
# It is already clean
````

````python
# Same for Scopus, check entries with missing keywords and check if there are proceedings of conferences in the Scopus query result
scopus_df_filtered = scopus_df[scopus_df['keywords'].isnull()]
print(f'No. of entries without keywords and abstract in Scopus query result: {scopus_df_filtered.shape[0]}')
print(scopus_df_filtered['title'].head())
print('-'*80)
scopus_df_proceedings = scopus_df[scopus_df['title'].str.contains('Proceedings', case=False, na=False)]
print(f'No. of entries with "Proceedings" in the title in Scopus query result: {scopus_df_proceedings.shape[0]}')
print(scopus_df_proceedings['title'].head())
print('-'*80)
# There are some proceedings, so we will remove all entries with "Proceedings" in the title from the Scopus query result
scopus_df_filtered = scopus_df[~scopus_df['title'].str.contains('Proceedings', case=False, na=False)]
print(f'No. of entries in Scopus query result after removing proceedings of conferences: {scopus_df_filtered.shape[0]}')
````

````python
# Reconciliation of the three datasets
slr_df = pd.concat(
    [acm_df_filtered[['title', 'keywords', 'abstract']],
     ieee_df[['title', 'keywords', 'abstract']], 
     scopus_df_filtered[['title', 'keywords', 'abstract']]],
    ignore_index=True)

print('Information about the SLR DataFrame')
print(slr_df.count())
print(slr_df.columns)
# Lets check if there are duplicated entries in the SLR DataFrame
slr_df_duplicates = slr_df[slr_df.duplicated(subset=['title'], keep=False)]
print(f'No. of duplicated entries in the SLR DataFrame: {slr_df_duplicates.shape[0]}')

slr_df_clean = slr_df.drop_duplicates(subset=['title'], keep='first')
print(f'No. of clean entries in the SLR DataFrame (after removing duplicates): {slr_df_clean.shape[0]}')

# Lets sample some entries from the SLR DataFrame
print(slr_df_clean[['title', 'abstract', 'keywords']].sample(5))
````

````python
# Store the list to a CSV file
slr_df_clean.to_csv('slr_data.csv', index=True)

# There are 270 entries in the CSV file. 
# Manual check to remove retracted papers, and/or entries that are not relevant for the SLR and preprunning was not able to remove them.
# Indeces to be removed: 4 (retracted), 108 (proceedings), 121 (proceedings), 124 (proceedings), 135 (proceedings), 137 (proceedings), 146 (proceedings)
slr_df_clean = slr_df_clean.drop(index=[4, 108, 121, 124, 135, 137, 146])
print(f'No. of clean entries in the SLR DataFrame after manual check: {slr_df_clean.shape[0]}')

# Overwrite the CSV file with the clean entries
slr_df_clean.reset_index(drop=True, inplace=True)
slr_df_clean.to_csv('slr_data.csv', index=True)
````
